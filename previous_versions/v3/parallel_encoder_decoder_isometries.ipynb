This notebook only records the two-isometry parallel realization and the basis numbering used for the matrices. It does not include the Choi/link-product verification or the entrywise comparison with the sequential comb.


## System numbering

- Encoder: `V_enc: P -> I1 I2 I3 R`.
- Encoder matrix rows are ordered as `|I1 I2 I3 R>`, with row index `8*I1 + 4*I2 + 2*I3 + R`; columns are ordered as `|P>`.
- Decoder: `V_dec: R O1 O2 O3 -> F A4`, where `A4` is a four-qubit ancillary output.
- Decoder matrix columns are ordered as `|R O1 O2 O3>`, with column index `8*R + 4*O1 + 2*O2 + O3`.
- Decoder matrix rows are ordered as `|F A4_1 A4_2 A4_3 A4_4>`, with row index `16*F + 8*A4_1 + 4*A4_2 + 2*A4_3 + A4_4`.


```python
import numpy as np

s2, s3, s6 = np.sqrt(2), np.sqrt(3), np.sqrt(6)
np.set_printoptions(precision=6, suppress=True, linewidth=160)


encoder_row_bits = ["I1", "I2", "I3", "R"]
encoder_col_bits = ["P"]
decoder_col_bits = ["R", "O1", "O2", "O3"]
decoder_row_bits = ["F", "A4_1", "A4_2", "A4_3", "A4_4"]

def bits_of(index, width):
    return tuple((index >> shift) & 1 for shift in range(width - 1, -1, -1))

encoder_row_numbering = {
    row: dict(zip(encoder_row_bits, bits_of(row, 4)))
    for row in range(16)
}
encoder_col_numbering = {
    col: {"P": col}
    for col in range(2)
}
decoder_col_numbering = {
    col: dict(zip(decoder_col_bits, bits_of(col, 4)))
    for col in range(16)
}
decoder_row_numbering = {
    row: dict(zip(decoder_row_bits, bits_of(row, 5)))
    for row in range(32)
}

basis_numbering = {
    "encoder_rows": encoder_row_numbering,
    "encoder_columns": encoder_col_numbering,
    "decoder_rows": decoder_row_numbering,
    "decoder_columns": decoder_col_numbering,
}
```
